```python
# Kütüphaneleri kuruyoruz
!pip install transformers torch -q

from transformers import pipeline

# Hugging Face'ten hazır bir Türkçe duygu analizi modeli yüklüyoruz
model_name = "savasy/bert-base-turkish-sentiment-cased"
classifier = pipeline("sentiment-analysis", model=model_name)

# İçinde karmaşık (mixed) duygular barındıran 15 örnek yorum
reviews = [
    "ürün kaliteli ama kargo çok geç geldi",
    "Telefonun kamerası harika fakat bataryası yarım günde bitiyor.",
    "Satıcı çok ilgiliydi ancak kargoda paket paramparça olmuş.",
    "Fiyatına göre muazzam bir kalite, kesinlikle tavsiye ederim.",
    "Rengi görseldeki gibi çok canlı ama kalıbı inanılmaz dar, iade.",
    "Malzemesi çok dandik, iki günde koptu, paranızı çöpe atmayın.",
    "Süpürge halıları çok iyi çekiyor ama çalışırken uçak motoru gibi ses çıkarıyor.",
    "Müşteri hizmetleri iade sürecinde çok zorluk çıkardı, bir daha buradan almam.",
    "Kargo ertesi gün elime ulaştı, paketleme de gayet sağlamdı.",
    "Kumaşı pamuklu ve yumuşacık fakat dikişleri sökük geldi.",
    "Hem fiyatı çok uygun hem de performansı beklediğimden çok yüksek.",
    "Ekran kalitesi muazzam ama arayüzde sürekli takılmalar yaşıyorum.",
    "Satıcı eksik parça göndermiş, mesajlara da asla cevap vermiyor.",
    "Tam bedenimi aldım, cuk oturdu, kumaşı da terletmiyor.",
    "Tasarımı çok şık duruyor ama boyutu beklediğimden çok daha küçük.",
]

print("--- Pretrained Model Trial ---")
for review in reviews:
    result = classifier(review)[0]
    print(f"Yorum: {review}")
    print(f"Tahmin: {result['label']} | Skor: {result['score']:.4f}\n")
```
